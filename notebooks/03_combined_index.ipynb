# Notebook 3 — Anomalies and combined index

**ADEWI** = 0.40 Z_rain + 0.35 Z_SM − 0.25 Z_LST  
**CADI** also includes NDVI/EVI (status index, not the early-warning index).

Classes: Mild < −0.5, Moderate < −1.0, Severe < −1.5, Extreme < −2.0.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))

from adewi.config import load_config
from adewi.data_loading import load_provincial_timeseries
from adewi.preprocessing import preprocess
from adewi.anomalies import compute_anomalies
from adewi.combined_index import compute_indices

cfg = load_config()
df, _ = load_provincial_timeseries(cfg)
clean, _ = preprocess(df)
zdf, clim = compute_anomalies(clean)
idx = compute_indices(zdf, cfg)
print(idx.groupby('province')[['adewi','cadi','z_ndvi']].describe())
print(idx['adewi_class'].value_counts())
```
