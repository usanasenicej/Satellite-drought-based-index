# Notebook 4 — Lead-time analysis (core contribution)

1. Detect NDVI drought events (Z_NDVI ≤ −1.0 for ≥ 2 composites).
2. Find the first ADEWI crossing of −1.0 in the preceding ~6 months.
3. Lead time (weeks) = (NDVI onset − ADEWI onset) / 7.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))

from adewi.config import load_config
from adewi.data_loading import load_provincial_timeseries
from adewi.preprocessing import preprocess
from adewi.anomalies import compute_anomalies
from adewi.combined_index import compute_indices
from adewi.lead_time import analyze_lead_time

cfg = load_config()
df, _ = load_provincial_timeseries(cfg)
clean, _ = preprocess(df)
zdf, _ = compute_anomalies(clean)
idx = compute_indices(zdf, cfg)
events, summary = analyze_lead_time(idx, cfg)
display(summary)
display(events)
```
